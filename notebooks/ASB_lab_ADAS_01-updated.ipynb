#### Python Implementation

The file integrity verification was implemented in Python using the standard libraries `hashlib` and `pathlib`.

The `pathlib` library was used to identify and access the original image files and their corresponding copies. The `hashlib` library was used to calculate SHA-256 hash values.

The verification procedure consisted of four steps:

1. Identify the corresponding image files in the original Sciebo dataset and the GitHub repository.
2. Read each image file in binary mode.
3. Calculate the SHA-256 hash of both versions.
4. Compare the hashes to determine whether the files are byte-for-byte identical.

The script processed four image pairs. All four comparisons returned matching SHA-256 values.

The complete implementation is stored in `verify_integrity.py`, while the corresponding image assessment and verification statuses are documented in `WP_A_image_quality_review.csv`.

The script is intended to provide a reproducible check of file-level consistency. It does not assess botanical classification accuracy, annotation correctness or the overall quality of the dataset.

## 1. Annotation Classes

For our sugar beet dataset, we used four classes for semantic segmentation:

- **Background (0):** Soil and other non-plant areas.
- **Sugar beet (1):** Visible sugar beet leaves and stems.
- **Weed (2):** Non-crop vegetation that we can confidently identify.
- **Ignore (255):** Areas where the plant identity or boundary is uncertain.

We annotated only the visible parts of the plants. When leaves overlap, we label the visible surfaces without guessing the hidden parts. We use the Ignore class for uncertain vegetation rather than assigning an incorrect label.

## 6.8 Dataset Limitations and Intended Use

The dataset was prepared to support research and practical experiments in semantic segmentation of sugar beet and non-crop vegetation.

However, several limitations should be considered.

First, the number of images from each source group is not balanced, and some images may contain similar visual information because they originate from closely related recordings.

Second, the weed class occupies a small proportion of the annotated pixels. This may limit the variety of non-crop vegetation represented in the dataset.

Third, some plant boundaries and small seedlings remain difficult to identify consistently. The Ignore class was used where the available visual evidence was insufficient.

The dataset can reasonably support experiments involving crop and weed segmentation under the conditions represented in the collected images. It should not be assumed to represent all agricultural environments, weed species or growth stages.
