## 5. Work package C: Annotate, review and assess consistency
**Read:** revisit Monarch [3], §7.1, and the annotation definitions in Weyler et al. [1].

Begin with a shared pilot of approximately **10 images**, independently annotated by every team member before discussing the results. Compare class assignments, missed vegetation and boundary placement. Retain the original annotations, record the disagreements, and agree corrections. Seek lecturer or domain-expert input when the issue is plant identification rather than interpretation of the guideline. Update the guideline before distributing the remaining images.

Produce **at least 100 distinct reviewed images or tiles**, with 120 as the target. Corrected pilot images count toward this total. Every image must be reviewed by a second team member. At least **20% of the final images** must have independent annotations from two members; the pilot can contribute to this subset. Review and independent annotation are different: a reviewer inspects existing labels, whereas an independent annotator works without seeing the first annotation.

Use overlays and a quantitative comparison to analyse agreement on the independently annotated subset **before adjudication**. For semantic masks, report class-wise intersection over union:

$$
\mathrm{IoU}_c = \frac{|A_c \cap B_c|}{|A_c \cup B_c|},
$$

where $A_c$ and $B_c$ contain the pixels assigned to class $c$ by the two annotators. Aggregate intersections and unions across the compared images before calculating each class result. Exclude pixels marked ignore by either annotator and report the excluded fraction separately. Report a class with zero union as not applicable, rather than perfect agreement. Keep crop and weed results separate; background agreement can dominate an overall pixel-accuracy score.

Discuss typical disagreements and the changes made to resolve them. Agreement is evidence of consistency, not proof of biological correctness. Do not claim a universal quality threshold from a small pilot.

**Evidence to include:** reviewed image count, review coverage, agreement results, annotation/review times and a small set of disagreement overlays with explanations. Retain annotator, reviewer and guideline-version information in your records.
